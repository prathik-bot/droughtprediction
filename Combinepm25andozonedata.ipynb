```python
import pandas as pd

# Load the PM2.5 dataset
pm25_df = pd.read_csv("Data/top_40_cities_pollution_data.csv", parse_dates=["Date"])

# Load the Ozone dataset
ozone_df = pd.read_csv("Data/top_40_cities_ozone_data.csv", parse_dates=["Date"])

# Merge on Date and Name only
merged_df = pd.merge(pm25_df, ozone_df[['Date', 'Name', 'Ozone']], on=["Date", "Name"], how="inner")

# Reorder columns: Date, Latitude, Longitude, PM2.5, Ozone, AQI, Name
desired_column_order = ["Date", "Latitude", "Longitude", "PM2.5", "Ozone", "AQI", "Name"]
merged_df = merged_df[desired_column_order]

# Save to a new CSV file
merged_df.to_csv("Data/top_40_cities_data.csv", index=False)

# Display first 5 rows where Name is 'San Jose'
san_jose_data = merged_df[merged_df["Name"] == "San Jose - Jackson"].head(5)
print(san_jose_data)

```
